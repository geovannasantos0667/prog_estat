```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#exponencial pelo método da inversão

M = 5000
lam = 3
exp = []

for i in range(M):
  U = np.random.uniform()  #gera uma uniforme
  X = -np.log(U)/lam #fórmula da inversa da exponencial
  exp.append(X)
print(exp)
```

```python
plt.hist(exp)
```

```python
plt.hist(np.random.exponential(1/3,1000))
```

```python
#aproximação de exponencial por geométrica

n = 10000
pn = lam/n  #parâmetro da geométrica
tn = []
for i in range(M):
  u = np.random.uniform()
  num = np.log(1-u)
  den = np.log(1-pn)
  x_geo = np.floor(num/den)+1
  tn.append(x_geo/n)
print(tn)
```

```python
plt.hist(tn)  #com n suficientemente grande a geométrica com parâmetro (lam/n) se aproxima da exponencial com parâmetro lam
```

```python
l = 2
pos =[]

for i in range(M):
  soma = 0  #vai somar as exponenciais
  cont = 0  #conta quantas exponenciais eu somei antes de dar um número maior que 1

  while True:
    uniforme = np.random.uniform()
    expo = -np.log(uniforme)/l

    if soma + expo > 1:
      break #se na primeira tentativa o resultado for maior que 1 o contador é 0

    soma = soma + expo
    cont += 1
  pos.append(cont)
print(pos)
```

```python
plt.hist(pos)
```
