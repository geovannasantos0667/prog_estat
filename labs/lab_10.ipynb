```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
#exercício 1
n = 10**4  #tamanho da amostra
X = np.random.uniform(size=n)  #amostra de uniformes
amostra = np.exp(-X**2)  #amostra de e**-x^2

esp = np.mean(amostra)
var = np.var(amostra)/n

ic = (esp + 2*np.sqrt(var), esp - 2*np.sqrt(var))

print(f'A variância do estimador é {var} e o intervalo de confiança do estimador é ({ic}).')
```

```python
#exercício 2
x = np.random.normal(0, 1, size=n)  #amostra da normal
ind = (x > 1)  #valores da normal maiores que 1

p_est = np.mean(ind)  #média
var_est = p_est*(1 - p_est)/n #variância
ic = (p_est - 2*np.sqrt(var_est), p_est + 2*np.sqrt(var_est))  #intervalo de confiança

print(p_est, var_est, ic)
```

```python
#exercício 3
x = np.random.uniform(size=n)
y = np.random.uniform(size=n)
ind = (x**2 + y**2 <= 1)

p_est = np.mean(ind)
pi_est = 4*p_est
var_est = 16*p_est*(1 - p_est)/n
ic = (pi_est - 2*np.sqrt(var_est), pi_est + 2*np.sqrt(var_est))

print(pi_est, var_est, ic)
```

```python
#exercício 4
x = np.random.uniform(size=n)

media_est = np.mean(x)
var_x_est = np.var(x, ddof=1)
var_media = var_x_est/n
ic = (media_est - 2*np.sqrt(var_media), media_est + 2*np.sqrt(var_media))

print(media_est, var_x_est, var_media, ic)
```

```python
#exercício 5
x = np.random.uniform(size=n)
y = np.random.uniform(size=n)
g = np.exp(-(x**2 + y**2))

I_est = np.mean(g)
var_est = np.var(g, ddof=1)/n
ic = (I_est - 2*np.sqrt(var_est), I_est + 2*np.sqrt(var_est))

print(I_est, var_est, ic)
```
