```python
#EXERCÍCIO 2.2
def bernoulli(p):
  U = np.random.uniform(size = 1000)
  X = []
  for i in range(1000):
    if U[i] < p:
      X.append(1)

    else:
      X.append(0)

  return X
```

```python
b = bernoulli(0.7) #amostra de bernoulli
media_teorica = 0.7 #média teórica
var_teorica = 0.21 #variância teórica
media = np.mean(b) #média amostral
var = np.var(b) #variância amostral
print(f'A média teórica é {media_teorica} e a média amostral é {media}, a variância teórica é {var_teorica} e a variãncia amostral é {var} ambos estão bem próximos')
```

```python
#EXERCÍCIO 2.3
def dado(A):  #tranforma valores uniformes em números de 1 a 6
  Y = []
  for i in range(1000):
    if 0 < A[i] < (1/6):
      Y.append(1)
    elif (1/6) < A[i] < (2/6):
      Y.append(2)
    elif (2/6) < A[i] < (3/6):
      Y.append(3)
    elif (3/6) < A[i] < (4/6):
      Y.append(4)
    elif (4/6) < A[i] < (5/6):
      Y.append(5)
    else:
      Y.append(6)
  return Y

print(dado(U)) #a função é aplicada direto na amostra uniforme
```

```python
def dado_1(A): #transforma valores uniformesem números de 1 a 6 sem usar if/elif/else
  Z = []
  mult = np.ceil(A*6) #uso a função teto para transformar em números de 1 a 6
  Z.append(mult)
  return Z

print(dado_1(U))
```

```python
A = np.random.uniform(size=1000)
def dado_2(A):  #transforma a uniforme em lançamentos de um dado viciado
  W = []
  for i in range(1000):
    if 0<A[i]<0.05:
      W.append(1)
    elif 0.05<A[i]<0.15:
      W.append(2)
    elif 0.15<A[i]<0.30:
      W.append(3)
    elif 0.30<A[i]<0.5:
      W.append(4)
      W.append(5)
    else:
      W.append(6)

  return W

print(dado_2(A))
```

```python
k = dado_2(A)
lista = {'1': [], '2': [], '3': [], '4': [], '5': [], '6': []}  #dicionário que separa os números de 1 a 6 em diferentes listas
for i in k:
  if k[i] == 1:
    lista['1'].append(i)
  elif k[i] == 2:
    lista['2'].append(i)
  elif k[i] == 3:
    lista['3'].append(i)
  elif k[i] == 4:
    lista['4'].append(i)
  elif k[i] == 5:
    lista['5'].append(i)
  else:
    lista['6'].append(i)

face1 = len(lista['1'])/1000 #frequência relativa das faces
face2 = len(lista['2'])/1000
face3 = len(lista['3'])/1000
face4 = len(lista['4'])/1000
face5 = len(lista['5'])/1000
face6 = len(lista['6'])/1000

print(f'\n Face 1: freq. relativa = {face1}, prob. teórica = 0.05.\n Face 2: freq.relativa = {face2}, prob.teórica = 0.10.\n Face 3: freq.relativa = {face3}, prob. teórica = 0.15.\n Face 4: freq. relativa = {face4}, prob. teórica = 0.20.\n Face 5: freq. relativa = {face5}, prob. teórica = 0.20.\n Face 6: freq.relativa = {face6}, pro. teórica = 0.30')
```

```python
#EXERCÍCIO 2.5
n = 50 #n° de lançamentos do dado
N = 1000 #quantidade de vezes que foi realizado o experimento
multinomial = []
face_1 = []
face_2 = []
face_3 = []
face_4 = []
for i in range(N):
  u = np.random.uniform(size=n) #gera uma amostra uniforme
  x1 = np.sum(u<0.10)
  x2 = np.sum((0.10<u)&(u<0.30))
  x3 = np.sum((0.30<u)&(u<0.60))
  x4 = np.sum(0.60<u)
  face_1.append(x1) #coloquei isso no código para poder calcular as médias mais pra frente
  face_2.append(x2)
  face_3.append(x3)
  face_4.append(x4)
  multinomial.append([x1,x2,x3,x4])]

multinomial = np.array(multinomial) #transforma uma lista em uma matriz para facilitar na hora de trabalhar com as colunas
plt.hist(multinomial, bins=10) #muda o agrupamento dos dados no gráfico
plt.show()


```

```python
#médias amostrais

med_x1 = np.mean(face_1)
med_x2 = np.mean(face_2)
med_x3 = np.mean(face_3)
med_x4 = np.mean(face_4)

#média teórica n*pi
esp_x1 = 50*0.10
esp_x2 = 50*0.20
esp_x3 = 50*0.30
esp_x4 = 50*0.40

print(f'para a face 1: média amostral {med_x1}, teórica {esp_x1}, para a face2 : média amostral {med_x2}, teórica {esp_x2}, para a face 3: média amostral {med_x3}, teórica {esp_x3}, para a face 4: média amostral {med_x4}, teórica {esp_x4}')
```

```python
#variância amostral
var_x1 = np.var(face_1)
var_x2 = np.var(face_2)
var_x3 = np.var(face_3)
var_x4 = np.var(face_4)

#variância teórica (n*pi)*(1-pi)
v_x1 = (50*0.10)*(1-0.10)
v_x2 = (50*0.20)*(1-0.20)
v_x3 = (50*0.30)*(1-0.30)
v_x4 = (50*0.40)*(1-0.40)

print(f'para a face 1: variância amostral {var_x1}, teórica {v_x1}, para a face 2: variância amostral {var_x2}, teórica {v_x2}, para a face 3: variância amostral {var_x3}, teórica {v_x3}, para a face 4: variância amostral {var_x4}, teórica {v_x4} ')
```

```python
#calcular covariância para duas faces distintas, escolhi face 1 e face 4

#cov amostral
cov = np.cov(face_1, face_4)[0,1] # sem esse [0,1] isso retornaria a matriz de covariancia
#covariância teórica -n*pi*pj
cova = -(50*0.10*0.40)

print(f'A covariância amostral entre X1 e X4 é {cov} e a covariância teórica é {cova}. Esse valor é negativo porque quando X1 aumenta a tendência é que X4 diminua.')
```
