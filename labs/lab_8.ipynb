```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

def f(x):
  return 2*(1/(2*np.pi)**0.5)*np.exp(-x**2/2) #f(x) normal

def g(x):
  return np.exp(-x) #g(x) exponencial

c =(f(1)/g(1)) #valor de c para o x ótimo x=1
x = np.linspace(-4,4,100000)
```

```python
n = 100000 #tamanho da amostra
amostra = []
sucessos = 0
tentativas = 0
while sucessos<n: #vai parar quando obtivermos o tamanho da amostra desejada
  U = np.random.uniform()
  y = -np.log(U)  #gero um y com distribuição exponencial
  u = np.random.uniform()
  if c*u*g(y)<=f(y):  #vê se esse ponto cai dentro da curva da f(x)
    amostra.append(y)  #se estiver dentro da curva f(x) adiciona o ponto da exponencial na amostra
    sucessos += 1
  tentativas +=1

print(amostra)
```

```python
plt.hist(amostra, density=True, bins=25)  #contruímos uma meia normal
plt.plot(x,f(x))
```

```python
p = 0.5  #probabilidade do sinal ser negativo ou positivo
amostra_2 = []
for i in amostra:
  uniforme = np.random.uniform()
  if uniforme<=p: #se for menor que p mantém o sinal
    amostra_2.append(i)
  else:
    amostra_2.append(-i) #se for maior que p troca o sinal
print(amostra_2)
```

```python
plt.hist(amostra_2, density=True,bins=25)
plt.plot(x,f(x)/2) #divide por dois porque queremos toda a distribuição normal
```

```python
plt.hist(np.random.normal(0,1, size=10000))  #distribuição normal gerada por numpy
plt.plot(x,f(x))
```

```python
#Suponha que queremos uma X com distribuição N(10,4), então o desvio padrão é 2
media = 10
sigma = 2
amostra_3 = []

for z in amostra_2:
  X = media + sigma*z  #isso por causa da normalização, se X~N(mi,var) então X-mi/var = Z~N(0,1)
  amostra_3.append(X)

print(amostra_3)
```

```python
plt.hist(amostra_3, density=True, bins=25)  #amostra de N(10,4)
```

```python
plt.hist(np.random.normal(10,2,size=100000),bins=25) #N(10,4), gerada por numpy
```

```python
p_aceitcao = sucessos/tentativas
taxa_teorica = 1/c
print(f'A probabilida de aceitação empírica é {p_aceitcao} e a probabilidade de aceitação teórica é {taxa_teorica}, ambas são compátiveis.')
```
