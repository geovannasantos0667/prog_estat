```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import time

#gerar uma distribuição geométrica através de uma bernoulli - método ingênuo
M = 1000  #tamanho da amostra
amostra = []  #para guardar a distribuição geométrica

# para calcular o tempo dos métodos
valores_p = np.linspace(0.1,0.99,100) #diferentes valores de p
metodo_i = []  #armazena o tempo para diferentes valores de p

for p in valores_p:
  inicio = time.time()
  for i in range(M):
    c = 1  #vai contar o número de lançamentos até o primeiro sucesso

    bern = np.random.uniform() < p

    while bern!= 1:
      bern = np.random.uniform() < p
      c +=1  #sempre que não ocorrer sucesso ele conta mais um lançamento
    amostra.append(c)
  fim = time.time()
  metodo_i.append(fim-inicio)

print(amostra)
```

```python
#quanto maior o p, menor será a quantidade de lançamentos, e quanto menor o p, maior é quantidade de lançamentos
plt.hist(amostra)
plt.hist(np.random.geometric(0.5,1000))  #são bem parecidos
```

```python
# método de inversão

geo = []

# para calcularo tempo do método
metodo_r = []

for p in valores_p:
  begin = time.time()
  for i in range(M):
    u = np.random.uniform()
    num = np.log(1-u) #númerador
    den = np.log(1-p) #denominador
    x = np.floor((num/den)+1) #calcula o x (quantidade de lançamentos até sucesso)
    geo.append(x)
  end = time.time()
  metodo_r.append(end-begin)

print(geo)
```

```python

plt.plot(
    valores_p,
    metodo_i,
    label = 'Metodo Ingênuo'
)

plt.plot(
    valores_p,
    metodo_r,
    label = 'Método de Inversão'
)

plt.xlabel('p')
plt.ylabel('Tempo(em segundos)')
plt.title('Comparação do Tempo dos Métodos')

plt.legend()
plt.grid()

plt.show()
```
