```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def f(x):
  return (20*x)*(1-x)**3  #densidade da X

x = np.linspace(0,1,1000)  #valores de x para criar uma constante
M = np.max(f(x))  #constante que deixa a função g acima do gráfico da X

def gera_var():
  U = 1  #densidade da uniforme
  y = np.random.uniform()  #número da nossa g(y)
  c = 0  #conta quantas tentativas até a primeira aceitação
  while U*M>f(y):
    U = np.random.uniform()
    y = np.random.uniform()
    c +=1
  return y,c


amostra = []
tentativas = []
for i in range(5000):
  amostra.append(gera_var()[0])
  tentativas.append(gera_var()[1])

print(amostra)
print(tentativas)
```

```python
plt.plot(x,f(x))
plt.axhline(y=M, c = 'red')
plt.hist(amostra, density=True, bins=50)
```

```python
prob_teo = 1/M  #probabilidade de aceitção do método AR
media_amostra = np.mean(tentativas)  #número médio de tentativas até a primeira aceitação
media_teo = 1/prob_teo
print(f'A média empírica é {media_amostra} e a média teórica é {media_teo}')
```
