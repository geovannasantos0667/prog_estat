```python
import matplotlib.pyplot as plt
import numpy as np
import time
```

```python
#Binomial negativa via Bernoulli
K = 100  #tamanho da amostra

p = 0.3  #probabilidade de sucesso
r = 4  #quantidade de sucessos desejado

valores_p = np.linspace(0.1,0.99,100)  #valores diferente de p para comparar os 3 métodos

X_ber = []
tempo_ber = []  #guarda os tempos de execução desse metodos para os diferente valores de p

for p in valores_p:
  inicio = time.time()

  for i in range(K):
    n = 0  #contador de ensaios
    s = 0  #número de sucessos
    while s<r:
      ber = np.random.uniform()<p
      n += 1

      if ber == 1:
        s += 1
    X_ber.append(n)
  fim = time.time()

  tempo_ber.append(fim-inicio)
print(X_ber)
```

```python
#Binomial negativa via soma de geometricas

X_geo = []
tempo_geo = []

for p in valores_p:
  start = time.time()

  for j in range(K):
    geo = []

    for i in range(r):
      u = np.random.uniform()
      num = np.log(1-u)
      den = np.log(1-p)
      x = np.floor(num/den)+1
      geo.append(x)
      y = np.sum(geo)
      X_geo.append(y)

  end = time.time()
  tempo_geo.append(end-start)
print(X_geo)
```

```python
#Binomial negativa via recursão

X_rec = []
tempo_rec = []

for p in valores_p:
  i  = time.time()

  for i in range(K):
    U = np.random.uniform()
    n = r
    pn = p**r
    F = pn
    while U > F:
      pn = pn*(n/(n-r+1))*(1-p)
      n += 1
      F = F + pn
    X_rec.append(n)

  f = time.time()
  tempo_rec.append(f-i)

print(X_rec)
```

```python
#Comparando o tempo dos 3 métodos

plt.plot(
    valores_p,
    tempo_ber,
    label = 'Aproximação Bernoulli'
)

plt.plot(
    valores_p,
    tempo_geo,
    label = 'Aproximação Geométrica'
)

plt.plot(
    valores_p,
    tempo_rec,
    label = 'Aproximação Recursão'
)

plt.xlabel('p')
plt.ylabel('Tempo(em segundos)')
plt.title('Comparação do Tempo dos Métodos para Binomial Negativa')

plt.legend()
plt.grid()

plt.show()
```

```python
#hipergeométrica

import numpy as np

def fisher_yates(M, N, n, K):
    amostras = []

    for _ in range(K):

        # Cria a população
        lista = [1] * N + [0] * M

        # Fisher-Yates: da frente para trás
        for i in range(n):

            U = np.random.uniform(0, 1)

            # escolhe uma posição entre i e o final
            j = i + int(np.floor((M + N - i) * U))

            # troca
            lista[i], lista[j] = lista[j], lista[i]

        # quantidade de 1's nos n elementos escolhidos
        X = sum(lista[:n])

        amostras.append(X)

    return amostras
```
