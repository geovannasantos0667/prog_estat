```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.special as special
```

Exercício 1

```python
n = 100_000
lam = 2
gamma = []
for i in range(n):
  exp = []
  for i in range(3):
    u = np.random.uniform()
    y = -np.log(u)/lam
    exp.append(y)
  gamma.append(np.sum(exp))  #X~Gamma(3,2)

amostra = []
for j in gamma:
  amostra.append(j**0.5)  #raiz de X, X~Gamma(3,2)

media = np.mean(amostra)
var = np.var(amostra)/n  #variância do estimador
ic = (media - 2*np.sqrt(var),media + 2*np.sqrt(var))
valor_teorico = 1.174982
print(f'O valor teórico {valor_teorico}, está contido no intervalo de confiança {ic}')
```

Exercício 2

```python
def h(a,b,x):
  return (1/special.beta(a,b))*x**(a-1)*(1-x)**(b-1)  #f(x)

def x_otimo(a,b):  #para a,b>1
  return (a-1)/(a+b-2)

c_otimo = h(2,5,x_otimo(2,5))
amostra_1 = []
sucessos = 0

while sucessos<n:
  y = np.random.uniform()
  uniforme = np.random.uniform()
  if uniforme <= h(2,5,y)/c_otimo:
    sucessos+=1
    amostra_1.append(y)  #Beta~(2,5) via aceitação-rejeição

contador = 0
for j in amostra_1:
  if j>0.4:
    contador+=1  #quantidade de X>0,4 na amostra

var = np.var(amostra_1)/n
p = contador/n
ic = (p - 2*np.sqrt(var), p + 2*np.sqrt(var))
print(f'O valor teórico 0.23328 está contido no intervalo de confiança {ic}')
```

Exercício 3

```python
def f(x):
  return (x**2 + 2*x + 1)
amostra_2 = []

#normal por box-muller
for i in range(n):
  u = np.random.uniform()*2*np.pi
  u2 = np.random.uniform()
  t = np.sqrt(-2*np.log(u2))*np.cos(u)
  x = 1 + 2*t  #mudei de N~(0,1) para N~(1,4)
  amostra_2.append(x)

integral = []
for j in amostra_2:
  integral.append(f(j))

media = np.mean(integral)
var = np.var(integral)/n
ic = (media - 2*np.sqrt(var), media + 2*np.sqrt(var))
print(f'O valor teórico da probabilidade, I=8, está contido no intervalo de confiança {ic}')
```

Exercício 4

```python
for i in range(n):
  exp = []
  for i in range(2):
    u = np.random.uniform()
    y = -np.log(u)
    exp.append(y)
  gamma.append(np.sum(exp))  #X~Gamma(2,1)

exponencial = []
for j in range(n):
  u = np.random.uniform()
  y = -np.log(u)
  exponencial.append(y)  #Y~Exp(1)

indicadora = []
for i in range(n):
  if gamma[i] > exponencial[i]:
    indicadora.append(1)
  else:
    indicadora.append(0)

media = np.mean(indicadora)
var = np.var(indicadora)/n
ic = (media - 2*np.sqrt(var), media + 2*np.sqrt(var))
print(f'O valor teórico 0.75 está contido no intervalo de confiaça {ic}')
```

Exercício 5

```python
def h(a,b,x):
  return (1/special.beta(a,b))*x**(a-1)*(1-x)**(b-1)  #f(x)

def x_otimo(a,b):  #para a,b>1
  return (a-1)/(a+b-2)

c_otimo = h(3,4,x_otimo(3,4))
amostra_3 = []
sucessos = 0

while sucessos<n:
  y = np.random.uniform()
  uniforme = np.random.uniform()
  if uniforme <= h(3,4,y)/c_otimo:
    sucessos+=1
    amostra_3.append(y)  #Beta~(3,4) via aceitação-rejeição

func = []
for i in amostra_3:
  func.append(np.exp(i))

I = special.beta(3,4)*np.mean(func)  #aproximação da integral por monte carlo
media = np.mean(func)
var = np.var(func)/n
ic = (I - 2*np.sqrt(var), I + 2*np.sqrt(var))
print(f'O intervalo de confiança de 95% é {ic}, e o I estimado por monte carlo é {I} ')
```

Exercício 6

```python
normal_1 = []
for i in range(n):
  u = np.random.uniform()*2*np.pi
  u2 = np.random.uniform()
  t = np.sqrt(-2*np.log(u2))*np.cos(u)  #X~N(0,1)
  normal_1.append(t)

normal_2 = []
for j in range(n):
  u = np.random.uniform()*2*np.pi
  u2 = np.random.uniform()
  v = np.sqrt(-2*np.log(u2))*np.sin(u)  #Y~N(0,1)  independentes
  normal_2.append(v)

indicadora = []
for k in range(n):
    if normal_1[k]**2 + normal_2[k]**2 > 5:  # P(X^2 + Y^2 > 5)
      indicadora.append(1)
    else:
      indicadora.append(0)

media = np.mean(indicadora)
var = np.var(indicadora)/n
ic = (media - 2*np.sqrt(var), media + np.sqrt(var))

print(f'Valor teórico: 0.082085, Valor do estimador: {media}, o ic é {ic}')
```

Exercício 7

```python
x1 = []
x2 = []
x3 = []
x4 = []
x5 = []

for i in range(n):
  u1 = np.random.uniform()
  u2 = np.random.uniform()
  u3 = np.random.uniform()
  u4 = np.random.uniform()
  u5 = np.random.uniform()
  y1 = -np.log(u1)/2
  y2 = -np.log(u2)/2
  y3 = -np.log(u3)/2
  y4 = -np.log(u4)/2
  y5 = -np.log(u5)/2
  x1.append(y1), x2.append(y2), x3.append(y3), x4.append(y4), x5.append(y5)  # Xi~Exp(2), Xi independentes

maximo = []
for j in range(n):
  maximo.append(max(x1[j], x2[j], x3[j], x4[j], x5[j]))  #M = max(X1,...,X5)

media = np.mean(maximo)
var = np.var(maximo)/n
ic = (media - 2*np.sqrt(var), media + 2*np.sqrt(var))

print(f'E[M] teórica é 1.141667 e o valor estimado por monte carlo é {media}, o ic é {ic}')
```

```python
normal = []
for i in range(n):
  u = np.random.uniform()*2*np.pi
  u2 = np.random.uniform()
  t = np.sqrt(-2*np.log(u2))*np.cos(u)  #X~N(0,1)
  normal.append(t)

amostra = []
for j in range(n):
  amostra.append(np.exp(normal[j]))  # X = e^Z

media = np.mean(amostra)
var = np.var(amostra)/n
ic = (media - 2*np.sqrt(var), media + 2*np.sqrt(var))

print(f'E[X] teórico é 1.648751 e por monte carlo é {media}, o ic é {ic}')

indicadora = []
for k in range(n):
  if amostra[k]>10:
    indicadora.append(1)  # P(X>10)
  else:
    indicadora.append(0)

media_indicadora = np.mean(indicadora)
var_indicadora = np.var(indicadora)/n
ic_indicadora = (media_indicadora - 2*np.sqrt(var_indicadora), media_indicadora + 2*np.sqrt(var_indicadora))

print(f'p teórico é 0.016511 e por monte carlo é {media_indicadora}, o ic é {ic_indicadora}')
```
