```python
import matplotlib.pyplot as plt
import numpy as np
import time
valores_p = np.linspace(0.1,0.99,100)
#gerando a normal

#pela uniforme
n1 = 1000 #tamanho da uniforme
media1 = 0.5
var1 = 1/12
N1 =[]
M = 5000

for i in range(M):

  u = np.random.uniform(size=n1)
  media_u = np.mean(u)

  Z_n1 = ((n1**0.5)*(media_u - media1))/var1**0.5
  N1.append(Z_n1)
print(N1)
```

```python
plt.hist(N1)
```

```python
#bernoulli

n2 = 1500
p = 0.5
media_2 = 0.5
var2 = 1/4
N2 = []
for i in range(M):
  ber = np.random.uniform(size=n2)<p
  media_ber = np.mean(ber)

  Z_n2 = ((n2**0.5)*(media_ber-media_2))/var2**0.5
  N2.append(Z_n2)
print(N2)
```

```python
plt.hist(N2)
```

```python
#geometrica
n3 = 100
media3 = 2
var3 = 2

N3 = []

for j in range(1000):
  geo = []
  for i in range(n3):
    u = np.random.uniform()
    num = np.log(1-u)
    den = np.log(1-p)
    x = np.floor((num/den))+1
    geo.append(x)
  media_geo = np.mean(geo)

  Z_n3 = ((n3**0.5)*(media_geo - media3))/var3**0.5
  N3.append(Z_n3)



```

```python
plt.hist(N3)
```

```python
#poisson
n4 = 100
media4 = 2
var4 = 2
N4 = []

for i in range(1000):
  P = []
  for i in range(n4):
    u = np.random.uniform()
    i = 0
    lam = 2
    p0 = np.e**(-lam)
    F = p0

    while u > F:
      i +=1
      p0 = (lam/i)*p0
      F = F + p0
    P.append(i)
  media_pos = np.mean(P)

  Z_n4 = ((n4**0.5)*(media_pos - media4))/var4**0.5
  N4.append(Z_n4)
print(N4)
```

```python
# aproximação de poisson por binomial de parametro lambida/n
pos = []
tempo_binomial = []
for p in valores_p:
  inicio = time.time()
  n = 50
  for i in range(M):
    uniforme = np.random.uniform(size=n)
    l = 2
    pn = l/n
    pos.append(np.sum(uniforme<pn))
  fim = time.time()
  tempo_binomial.append(fim-inicio)
print(pos)
```

```python
plt.hist(pos)
```

```python
plt.hist(np.random.poisson(2,size=5000))
```

```python
#poisson método recursivo
tempo_inversao = []
for p in valores_p:
  begin = time.time()
  for i in range(M):
    u = np.random.uniform()
    i = 0
    lam = 2
    p0 = np.e**(-lam)
    F = p0

    while u > F:
      i +=1
      p0 = (lam/i)*p0
      F = F + p0
    P.append(i)
  end = time.time()
  tempo_inversao.append(end-begin)
```

```python
plt.hist(P)
```

```python
plt.plot(
    valores_p,
    tempo_binomial,
    label = 'Metodo Binomial'
)

plt.plot(
    valores_p,
    tempo_inversao,
    label = 'Método de Inversão'
)

plt.xlabel('p')
plt.ylabel('Tempo(em segundos)')
plt.title('Comparação do Tempo dos Métodos')

plt.legend()
plt.grid()

plt.show()
```
