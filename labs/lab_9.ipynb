```python
import matplotlib.pyplot as plt
import numpy as np
import scipy.special as special
```

```python
#exercício 1
k = 100 #parâmetro da gama
N = 1000
lam = 3
gama = []

for i in range(N):
  amostra = []
  for i in range(k):
    u = np.random.uniform()
    exp = -np.log(u)/lam
    amostra.append(exp)
  gama.append(np.sum(amostra))

print(gama)
```

```python
plt.hist(gama, density=True)  #gamma via soma de exponenciais
```

```python
plt.hist(np.random.gamma(100,1/3,1000))  #gamma gerada por numpy
```

```python
#exercício 2
a = 1.5
mi = lam/a
amostra_2 = []

def f(x):
  return (lam**a/special.gamma(a))*x**(a-1)*np.exp(-lam*x)


def g(x):
  return mi*np.exp(-mi*x)

x_otimo = (a-1)/(lam-mi)
c = f(x_otimo)/g(x_otimo)

for i in range(N):
  U = np.random.uniform()
  Y = -np.log(U)/mi
  unif = np.random.uniform()
  if unif <= f(Y)/(c*g(Y)):
    amostra_2.append(Y)
```

```python
plt.hist(amostra_2, bins=25)  #gamma gerada via aceitação-rejeição por Exponencial
```

```python
plt.hist(np.random.gamma(1.5,1/3,1000), bins=25)  #gamma gerada por numpy
```

```python
plt.hist(X, bins=25)  #gamma via resultado visto em sala
```
